A notebook is a page of boxes called **cells**. There are two kinds:

- **Text cells**, like this one, hold notes for people. Colab doesn't run them.
- **Code cells** have a grey background and a ▶ button on their left. They hold Python. When you run one, Colab works it out and shows the result underneath.

Work down the page in order. At the end, a cell prints a code to type on the operations board.

## 1. Run a cell
Click into the code cell below, then press **Shift + Enter**, or click its ▶ button. The first run takes a few seconds while Colab connects. If Colab warns that this notebook wasn't written by Google, choose **Run anyway**.

```python
print("Station K, reporting in")
```

The result appears under the cell. Shift + Enter also moves you down to the next cell, so you can keep pressing it to work down the notebook.

## 2. Store two numbers
A code cell can store a value under a name, so that later cells can use it. Run this cell. It stores the number of calls this week and the number of crews sent to each call, and shows nothing.

```python
calls = 14
crews = 2
```

## 3. Use them in another cell
This cell multiplies the two numbers you stored, to get the crew turnouts this week. Run it. You should see **28**.

```python
calls * crews
```

When the last line of a code cell gives a value, Colab shows it under the cell.

## 4. Change the code, then run it again
The Data Office has corrected the figure: each call got **4** crews, not 2.

1. Go back to the cell in step 2 and change `crews = 2` to `crews = 4`.
2. Run that cell again.
3. Run the cell in step 3 again. It should now show **56**.

Editing code changes nothing on its own. Colab only works out a result when you run the cell, so a result can be out of date until you do.

## 5. Add your own cells
Move your mouse just below this text cell. Two buttons appear: **+ Code** and **+ Text**.

1. Click **+ Text** and write a note, such as *Crews changed from 2 to 4, corrected by the Data Office.* Click outside the cell to finish. In an operation, notes like this record what you did to the data and why.
2. Click **+ Code**, type `calls + crews` and run it. You should see **18**.

## 6. When Colab resets
If you leave a notebook alone for about an hour and a half, or close the tab, Colab resets. Make it happen now: choose **Runtime → Restart session**, and confirm.

Then run **only** the cell below. It fails with **NameError**, because Colab has forgotten `calls` and `crews`: nothing has run since the reset. Your code is all still on the page. Only the results and the stored values are gone.

```python
print(calls, crews)
```

**The fix:** choose **Runtime → Run all**. Colab runs every cell from the top, so `calls` and `crews` are stored again, and the cell above works.
