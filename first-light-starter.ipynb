## 1. Look at the file, then load it
The file is on the Field Manual site. Before pandas reads it, look at the top of it exactly as it is, the way you'd read a document before acting on it. A file can start with a title, notes or blank lines that aren't data.

```
url = "https://station-k.vercel.app/data/practice/practice_calls.csv"

import requests
print(requests.get(url).text[:500])   # the first 500 characters, as they are in the file
```

This file starts straight with the column names, so pandas can read it as it is. Write down where your data came from: your report will need it.

```
calls = pd.read_csv(url)
```

## 2. Look at it

```
calls.head()
```

```
len(calls)
```

## 3. Average by group
The average response time for each station, worst first:

```
calls.groupby("station")["response_min"].mean().sort_values(ascending=False).round(2)
```

Read the result before you use it. Osprey's average is below zero, which no response time can be, and `kestrel` shows up as a separate station from `Kestrel`. An average is only as good as the rows that go into it.
